```python
# Convert the original date column to text first
date_text = transactions_clean["TransactionDate"].astype("string").str.strip()

# Try DD/MM/YYYY format
date_dmy = pd.to_datetime(
    date_text,
    format="%d/%m/%Y",
    errors="coerce"
)

# Try YYYY-MM-DD for anything that wasn't converted
date_ymd = pd.to_datetime(
    date_text,
    format="%Y-%m-%d",
    errors="coerce"
)

# Combine the two results
transactions_clean["TransactionDate"] = date_dmy.fillna(date_ymd)
```

```python
def convert_date(x):
    if pd.isna(x):
        return pd.NaT
    
    x = str(x).strip()
    
    for fmt in ["%d/%m/%Y", "%Y-%m-%d"]:
        try:
            return pd.to_datetime(x, format=fmt)
        except ValueError:
            pass
    
    return pd.NaT


transactions_clean["TransactionDate"] = (
    transactions_clean["TransactionDate"].apply(convert_date)
)
```
